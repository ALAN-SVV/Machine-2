### División del conjunto de datos

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y)
print("Tamaño de entrenamiento:", len(X_train))
print("Tamaño de prueba:", len(X_test))
```

## Implementación de un aMP Neurona más avanzada

```python
# Importamos las bibliotecas de numpy
import numpy as np
from sklearn.metrics import accuracy_score

class MPNeurona:
    # Función constructor
    def __init__(self):
        self.threshold = None

    # Creamos una función para el modelo
    def model(self, x):
        # Función escalón
        # ...
        return (x >= self.threshold)

    # Creamos la predicción
    def predict(self, X):
        Y = []
        for x in X:
            result = self.model(x)
            Y.append(result)
        return np.array(Y)

    # Creamos el método para encontrar el mejor threshold
    def fit(self, X, Y):
        accuracy = {}
        for i in range(X.shape[1]):
            for threshold in X[:, i]:
                self.threshold = threshold
                Y_pred = self.predict(X[:, i])
                accuracy[threshold] = accuracy_score(Y_pred, Y)
        # Seleccionamos el threshold que mejore nuestras predicciones
        self.threshold = max(accuracy, key=accuracy.get)
        return self
```

Los datos de entrada siguen teniendo problemas debido a que el conjunto de datos las características de entrada contienen valores continuos, sin embargo, la neurona MP solo procesa características de entrada con valores binarios.

```python
# Antes transformamos los valores en un valor binario
import matplotlib.pyplot as plt

plt.hist(X[:,1], X[:,1], bins=X[:,1])
plt.show()
```

```python
# Visualizamos las características de entrada para el valor binario
X_train_bin = pd.get_dummies(pd.DataFrame(X_train, columns=cancer_data.feature_names))
X_test_bin  = pd.get_dummies(pd.DataFrame(X_test,  columns=cancer_data.feature_names))
X_train_bin
```

```python
# Entrenamos el modelo MP
mp_neurona = MPNeurona()
mp_neurona = mp_neurona.fit(X_train, y_train)

# Obtenemos el threshold
print("Mejor threshold:", mp_neurona.threshold)
```

```python
# Realizamos la predicción por una clase de prueba
Y_pred = mp_neurona.predict(X_test_bin.to_numpy())
Y_pred
```

```python
# Realizamos la comparación de nuestras predicciones
print(accuracy_score(Y_pred, y_test))
```
